## Part 1. Find your data

Your project needs something for the model to know that it was not trained on.
Which kind depends on what you are building:

- **An agent or a RAG system** needs a **dataset or document collection** to
  retrieve from: a set of PDFs, a spreadsheet, a website, an API, a database.
- **A fine-tuning project** needs a **corpus**: many examples of the text you
  want the model to produce, in a consistent format.

Places worth searching:

- **Hugging Face Datasets** — huggingface.co/datasets. Searchable, and most load
  in one line of code.
- **Kaggle** — kaggle.com/datasets. Free account needed to download.
- **Government and public data** — data.gov, the Census Bureau, state open-data
  portals, the Library of Congress.
- **Your own sources** — course documents, a club's records, a public website,
  a department's handbook. If you did not make it, ask whoever did.
- **An API** — many services (weather, sports, finance, news) expose their data
  this way. Read the terms; some require a key.

## Part 3. Pick your agentic RAG design

Go back to Tuesday's reading, **Section 5: Taxonomy of Agentic RAG Systems**. It
describes seven kinds of system:

| Section | System | In one line |
|:--|:--|:--|
| 5.1 | Single-agent router | One agent decides which source to search, if any |
| 5.2 | Multi-agent | Several specialist agents, each with its own source or tool |
| 5.3 | Hierarchical | A top-level agent delegates to agents below it |
| 5.4 | Corrective | Checks what it retrieved and searches again if it is poor |
| 5.5 | Adaptive | Decides how much retrieval a question needs — none, one, or many |
| 5.6 | Graph-based | Retrieves over a knowledge graph of entities and relations |
| 5.7 | Agentic document workflows | Works through documents as a multi-step process |

Pick the one that best fits **your** project and **your** data from Part 1. Then
read **Section 10** before you finalize — it argues that agentic RAG is not always
the right default. If the plain pipeline you built in Week 10 is the honest
answer for your project, that is a legitimate choice, and arguing for it well
scores as highly as any other.
